```python
import pandas as pd
import numpy as np

# main_df = lastmile_events_cust.copy()
main_df = lastmile_events.copy()

# Filter
asn_inc = ["FM", "RC", "OV", "TS"]
del_t = ["REG"]

main_df = main_df.loc[
    main_df["ASN_TYPE_CD"].isin(asn_inc)
    & main_df["DELIVERY_TYPE"].isin(del_t)
].copy()

required_columns = [
    "ASN_TYPE_CD",
    "DELIVERY_TYPE",
    "ROUTE_ID",
    "DC_CD",
    "FILL_DC_CD",
    "DLVRY_ACTL_DATETIME",
    "DELIVERY_SEQ",
    "CNSLDTN_ID",
    "TOTAL_RATE",
    "LASTMILE_TOTAL_COST",
    "TOPSIDE_COST"
]

missing_columns = [
    col for col in required_columns
    if col not in main_df.columns
]

if missing_columns:
    raise KeyError(
        f"Missing required columns: {missing_columns}"
    )

# Convert calculation columns
main_df["DLVRY_ACTL_DATETIME"] = pd.to_datetime(
    main_df["DLVRY_ACTL_DATETIME"],
    errors="coerce"
)

main_df["DELIVERY_SEQ"] = pd.to_numeric(
    main_df["DELIVERY_SEQ"],
    errors="coerce"
)

main_df["TOTAL_RATE"] = pd.to_numeric(
    main_df["TOTAL_RATE"],
    errors="coerce"
).fillna(0.0)

main_df["LASTMILE_TOTAL_COST"] = pd.to_numeric(
    main_df["LASTMILE_TOTAL_COST"],
    errors="coerce"
).fillna(0.0)

main_df["TOPSIDE_COST"] = pd.to_numeric(
    main_df["TOPSIDE_COST"],
    errors="coerce"
).fillna(0.0)

# Route-level key:
# ROUTE_ID | DC_CD | delivery date
main_df["helper column"] = (
    main_df["ROUTE_ID"].astype("string").fillna("MISSING_ROUTE")
    + "|"
    + main_df["DC_CD"].astype("string").fillna("MISSING_DC")
    + "|"
    + main_df["DLVRY_ACTL_DATETIME"]
    .dt.strftime("%Y%m%d")
    .fillna("MISSING_DATE")
)

_dlvry_dt = main_df["DLVRY_ACTL_DATETIME"]

main_df["delivery date"] = _dlvry_dt.dt.normalize()
main_df["year"] = _dlvry_dt.dt.year
main_df["month"] = _dlvry_dt.dt.month

# Stop helper:
# ROUTE_ID | DC_CD | delivery date | DELIVERY_SEQ
stop_num = main_df["DELIVERY_SEQ"]

main_df["stop helper"] = (
    main_df["helper column"]
    + "|"
    + np.where(
        stop_num.fillna(0).eq(0),
        "RETURN",
        stop_num.fillna(0).astype(int).astype(str)
    )
)

# Distinct route stop flag
main_df["dist stop flag"] = (
    ~main_df["stop helper"].duplicated(keep="first")
).astype(int)

main_df.loc[
    main_df["stop helper"].eq(""),
    "dist stop flag"
] = 0

# Total Route Charge per route
rc_total = (
    main_df.loc[
        main_df["ASN_TYPE_CD"].eq("RC")
    ]
    .groupby(
        "helper column",
        dropna=False
    )["TOTAL_RATE"]
    .sum()
)

# Number of distinct FM stops per route
fm_cnt = (
    main_df.loc[
        main_df["ASN_TYPE_CD"].eq("FM")
        & main_df["dist stop flag"].eq(1)
    ]
    .groupby(
        "helper column",
        dropna=False
    )
    .size()
)

main_df["_rc_total"] = (
    main_df["helper column"]
    .map(rc_total)
    .fillna(0.0)
)

main_df["_fm_cnt"] = (
    main_df["helper column"]
    .map(fm_cnt)
    .fillna(0)
)

# Route has a Route Charge
_has_rc = main_df["_rc_total"].gt(0)

# Original cost-per-stop calculation before TS
main_df["cost per stop"] = np.where(
    main_df["ASN_TYPE_CD"].isin(["RC", "TS"]),
    0.0,
    np.where(
        ~_has_rc,
        np.where(
            main_df["ASN_TYPE_CD"].eq("OV"),
            main_df["LASTMILE_TOTAL_COST"],
            main_df["TOTAL_RATE"]
        ),
        np.where(
            main_df["ASN_TYPE_CD"].eq("FM")
            & main_df["dist stop flag"].eq(1),
            main_df["TOTAL_RATE"]
            + np.where(
                main_df["_fm_cnt"].gt(0),
                main_df["_rc_total"] / main_df["_fm_cnt"],
                0.0
            ),
            np.where(
                main_df["ASN_TYPE_CD"].eq("FM")
                & main_df["dist stop flag"].eq(0),
                0.0,
                main_df["TOTAL_RATE"]
            )
        )
    )
)

# Preserve CPS before topside
main_df["cost_per_stop_excl_ts"] = (
    main_df["cost per stop"]
)

# Monday-Sunday TS week
main_df["ts week start"] = (
    main_df["delivery date"]
    - pd.to_timedelta(
        main_df["delivery date"].dt.weekday,
        unit="D"
    )
)

main_df["ts week end"] = (
    main_df["ts week start"]
    + pd.Timedelta(days=6)
)

main_df["ts week"] = (
    main_df["ts week start"].dt.strftime("%Y-%m-%d")
    + " / "
    + main_df["ts week end"].dt.strftime("%Y-%m-%d")
)

# TS calculation key:
# DC_CD | Monday-Sunday week
main_df["ts key"] = (
    main_df["DC_CD"].astype("string").fillna("MISSING_DC")
    + "|"
    + main_df["ts week start"]
    .dt.strftime("%Y%m%d")
    .fillna("MISSING_WEEK")
)

# TOPSIDE_COST is one weekly charge per DC_CD.
#
# max() captures the one-time weekly charge without multiplying
# it if the weekly value appears on more than one TS source row.
ts_total = (
    main_df.loc[
        main_df["ASN_TYPE_CD"].eq("TS")
        & main_df["DC_CD"].notna()
        & main_df["ts week start"].notna()
    ]
    .groupby(
        "ts key",
        dropna=False
    )["TOPSIDE_COST"]
    .max()
)

# Eligible delivery records for the TS denominator
#
# A TS stop:
# - is an FM or OV delivery
# - has DELIVERY_SEQ > 0
# - has a CNSLDTN_ID
# - belongs to a valid DC_CD and delivery day
main_df["_ts_stop_eligible"] = (
    main_df["ASN_TYPE_CD"].isin(["FM", "OV"])
    & main_df["DELIVERY_SEQ"].gt(0)
    & main_df["CNSLDTN_ID"].notna()
    & main_df["DC_CD"].notna()
    & main_df["delivery date"].notna()
    & main_df["ts week start"].notna()
)

# Daily consolidation-stop key:
# DC_CD | delivery date | CNSLDTN_ID
#
# The same CNSLDTN_ID:
# - repeated on the same day counts once
# - appearing on another day counts again
main_df["ts daily cons key"] = pd.Series(
    pd.NA,
    index=main_df.index,
    dtype="string"
)

main_df.loc[
    main_df["_ts_stop_eligible"],
    "ts daily cons key"
] = (
    main_df.loc[
        main_df["_ts_stop_eligible"],
        "DC_CD"
    ].astype("string")
    + "|"
    + main_df.loc[
        main_df["_ts_stop_eligible"],
        "delivery date"
    ].dt.strftime("%Y%m%d")
    + "|"
    + main_df.loc[
        main_df["_ts_stop_eligible"],
        "CNSLDTN_ID"
    ].astype("string")
)

# First row for each distinct daily CNSLDTN_ID
main_df["ts daily distinct stop flag"] = (
    main_df["_ts_stop_eligible"]
    & ~main_df.duplicated(
        subset=[
            "DC_CD",
            "delivery date",
            "CNSLDTN_ID"
        ],
        keep="first"
    )
).astype(int)

# Daily distinct CNSLDTN_ID count by DC_CD
ts_daily_stop_count = (
    main_df.loc[
        main_df["_ts_stop_eligible"]
    ]
    .groupby(
        [
            "DC_CD",
            "delivery date",
            "ts week start",
            "ts key"
        ],
        dropna=False
    )["CNSLDTN_ID"]
    .nunique()
    .reset_index(
        name="daily_distinct_stop_count"
    )
)

# Weekly stop count:
# sum of each day's distinct CNSLDTN_ID count
#
# Example:
# Monday distinct cons IDs = 20
# Tuesday distinct cons IDs = 22
# Wednesday distinct cons IDs = 18
# Weekly denominator = 60
ts_weekly_stop_count = (
    ts_daily_stop_count
    .groupby(
        "ts key",
        dropna=False
    )["daily_distinct_stop_count"]
    .sum()
)

# Map one-time weekly TS charge and weekly stop denominator
main_df["_ts_total"] = (
    main_df["ts key"]
    .map(ts_total)
    .fillna(0.0)
)

main_df["_ts_cnt"] = (
    main_df["ts key"]
    .map(ts_weekly_stop_count)
    .fillna(0)
)

# TS CPS:
# one-time weekly TOPSIDE_COST
# divided by sum of daily distinct CNSLDTN_ID counts
main_df["ts_rate_per_stop"] = np.where(
    main_df["_ts_cnt"].gt(0),
    main_df["_ts_total"] / main_df["_ts_cnt"],
    0.0
)

# Identify one existing CPS-bearing row for each daily distinct cons ID.
#
# This prevents the TS charge being added more than once when the
# same CNSLDTN_ID has multiple source records on the same day.
main_df["ts cps stop flag"] = 0

ts_cps_candidate_mask = (
    main_df["_ts_stop_eligible"]
    & main_df["cost_per_stop_excl_ts"].ne(0)
)

main_df.loc[
    ts_cps_candidate_mask,
    "ts cps stop flag"
] = (
    ~main_df.loc[
        ts_cps_candidate_mask
    ]
    .duplicated(
        subset=[
            "DC_CD",
            "delivery date",
            "CNSLDTN_ID"
        ],
        keep="first"
    )
).astype(int)

# TS cost is shown on the CPS row representing each daily distinct stop
main_df["ts_cost_per_stop"] = np.where(
    main_df["ts cps stop flag"].eq(1)
    & main_df["_ts_total"].gt(0)
    & main_df["_ts_cnt"].gt(0),
    main_df["ts_rate_per_stop"],
    0.0
)

# Final cost per stop including TS
main_df["cost per stop"] = (
    main_df["cost_per_stop_excl_ts"]
    + main_df["ts_cost_per_stop"]
)
```

```python
# Daily TS stop-count detail
ts_daily_stop_count = (
    ts_daily_stop_count
    .merge(
        ts_total.rename("total_ts_charge"),
        left_on="ts key",
        right_index=True,
        how="left"
    )
)

ts_daily_stop_count["total_ts_charge"] = (
    ts_daily_stop_count["total_ts_charge"]
    .fillna(0.0)
)

ts_daily_stop_count["weekly_stop_count"] = (
    ts_daily_stop_count["ts key"]
    .map(ts_weekly_stop_count)
    .fillna(0)
)

ts_daily_stop_count["ts_cost_per_stop"] = np.where(
    ts_daily_stop_count["weekly_stop_count"].gt(0),
    (
        ts_daily_stop_count["total_ts_charge"]
        / ts_daily_stop_count["weekly_stop_count"]
    ),
    0.0
)

ts_daily_stop_count = (
    ts_daily_stop_count[
        [
            "DC_CD",
            "delivery date",
            "ts week start",
            "daily_distinct_stop_count",
            "weekly_stop_count",
            "total_ts_charge",
            "ts_cost_per_stop"
        ]
    ]
    .sort_values(
        [
            "DC_CD",
            "ts week start",
            "delivery date"
        ]
    )
    .reset_index(drop=True)
)

# All routes included in each DC_CD/week
ts_routes = (
    main_df.loc[
        main_df["_ts_stop_eligible"]
    ]
    .groupby(
        [
            "DC_CD",
            "ts week start"
        ],
        dropna=False
    )["ROUTE_ID"]
    .agg(
        lambda routes: ", ".join(
            sorted(
                routes
                .dropna()
                .astype(str)
                .unique()
                .tolist()
            )
        )
    )
    .reset_index(
        name="all_routes"
    )
)

# Route count by DC_CD/week
ts_route_count = (
    main_df.loc[
        main_df["_ts_stop_eligible"]
    ]
    .groupby(
        [
            "DC_CD",
            "ts week start"
        ],
        dropna=False
    )["ROUTE_ID"]
    .nunique()
    .reset_index(
        name="route_count"
    )
)

# Weekly TS validation
#
# Sum ts_cost_per_stop only from the selected daily distinct-stop rows.
ts_validation = (
    main_df.loc[
        main_df["ts cps stop flag"].eq(1)
    ]
    .groupby(
        [
            "DC_CD",
            "ts week start",
            "ts week",
            "ts key"
        ],
        dropna=False
    )
    .agg(
        total_ts_charge=(
            "_ts_total",
            "first"
        ),
        weekly_stop_count=(
            "_ts_cnt",
            "first"
        ),
        ts_cost_per_stop=(
            "ts_rate_per_stop",
            "first"
        ),
        allocated_ts_charge=(
            "ts_cost_per_stop",
            "sum"
        )
    )
    .reset_index()
)

ts_validation["allocation_difference"] = (
    ts_validation["allocated_ts_charge"]
    - ts_validation["total_ts_charge"]
)

ts_validation["allocation_balanced"] = np.isclose(
    ts_validation["allocated_ts_charge"],
    ts_validation["total_ts_charge"],
    rtol=1e-9,
    atol=0.01
)

ts_all_routes_df = (
    ts_validation
    .merge(
        ts_routes,
        on=[
            "DC_CD",
            "ts week start"
        ],
        how="left"
    )
    .merge(
        ts_route_count,
        on=[
            "DC_CD",
            "ts week start"
        ],
        how="left"
    )
)

ts_all_routes_df = (
    ts_all_routes_df[
        [
            "DC_CD",
            "ts week",
            "all_routes",
            "route_count",
            "total_ts_charge",
            "weekly_stop_count",
            "ts_cost_per_stop",
            "allocated_ts_charge",
            "allocation_difference",
            "allocation_balanced"
        ]
    ]
    .rename(
        columns={
            "weekly_stop_count": "total_weekly_stops"
        }
    )
    .sort_values(
        [
            "DC_CD",
            "ts week"
        ]
    )
    .reset_index(drop=True)
)

# One row per daily distinct cons ID carrying CPS and TS
ts_stop_detail = (
    main_df.loc[
        main_df["ts cps stop flag"].eq(1),
        [
            "DC_CD",
            "ts week",
            "delivery date",
            "ROUTE_ID",
            "DLVRY_ACTL_DATETIME",
            "DELIVERY_SEQ",
            "CNSLDTN_ID",
            "ASN_TYPE_CD",
            "_ts_total",
            "_ts_cnt",
            "cost_per_stop_excl_ts",
            "ts_rate_per_stop",
            "ts_cost_per_stop",
            "cost per stop"
        ]
    ]
    .rename(
        columns={
            "_ts_total": "total_ts_charge",
            "_ts_cnt": "total_weekly_stops"
        }
    )
    .sort_values(
        [
            "DC_CD",
            "ts week",
            "delivery date",
            "ROUTE_ID",
            "DELIVERY_SEQ",
            "CNSLDTN_ID"
        ]
    )
    .reset_index(drop=True)
)

# Derived stop counts
#
# A delivery stop is a distinct DELIVERY_SEQ > 0
# on an FM/OV row.
_seq = pd.to_numeric(
    main_df["DELIVERY_SEQ"],
    errors="coerce"
)

main_df["delivery stop flag"] = (
    main_df["ASN_TYPE_CD"].isin(["FM", "OV"])
    & _seq.gt(0)
    & ~main_df.duplicated(
        subset=[
            "helper column",
            "DELIVERY_SEQ"
        ],
        keep="first"
    )
).astype(int)

# Distinct stops per route
main_df["route_stop_count"] = (
    main_df.groupby(
        "helper column",
        dropna=False
    )["delivery stop flag"]
    .transform("sum")
)

# Distinct stops per fill DC per month
main_df["fill_dc_stop_count"] = (
    main_df.groupby(
        [
            "FILL_DC_CD",
            "year",
            "month"
        ],
        dropna=False
    )["delivery stop flag"]
    .transform("sum")
)

# TS source-data validation
#
# max(TOPSIDE_COST) is used as the one-time weekly charge.
ts_source_check = (
    main_df.loc[
        main_df["ASN_TYPE_CD"].eq("TS")
    ]
    .groupby(
        [
            "DC_CD",
            "ts week"
        ],
        dropna=False
    )
    .agg(
        ts_source_rows=(
            "TOPSIDE_COST",
            "size"
        ),
        ts_nonzero_rows=(
            "TOPSIDE_COST",
            lambda values: values.ne(0).sum()
        ),
        ts_unique_values=(
            "TOPSIDE_COST",
            "nunique"
        ),
        actual_weekly_ts_charge=(
            "TOPSIDE_COST",
            "max"
        ),
        summed_ts_for_comparison=(
            "TOPSIDE_COST",
            "sum"
        )
    )
    .reset_index()
)

# Remove TS source rows after allocation
main_df = main_df.loc[
    main_df["ASN_TYPE_CD"].ne("TS")
].copy()

# Remove internal calculation columns
#
# Retained:
# delivery date
# year
# month
# ts week
# ts week start
# ts week end
# cost_per_stop_excl_ts
# ts_rate_per_stop
# ts_cost_per_stop
# cost per stop
# route_stop_count
# fill_dc_stop_count
main_df.drop(
    columns=[
        "_rc_total",
        "_fm_cnt",
        "_ts_total",
        "_ts_cnt",
        "_ts_stop_eligible",
        "helper column",
        "stop helper",
        "dist stop flag",
        "delivery stop flag",
        "ts key",
        "ts daily cons key",
        "ts daily distinct stop flag",
        "ts cps stop flag"
    ],
    inplace=True,
    errors="ignore"
)

main_df.reset_index(
    drop=True,
    inplace=True
)

# # Final row-level data
# display(main_df.head())

# # Daily distinct CNSLDTN_ID counts contributing to each weekly denominator
# display(ts_daily_stop_count)

# # Weekly summary by DC_CD with all routes
# display(ts_all_routes_df)

# # One row per daily distinct CNSLDTN_ID stop
# display(ts_stop_detail)

# # Weekly source-charge check
# display(
#     ts_source_check.sort_values(
#         [
#             "DC_CD",
#             "ts week"
#         ]
#     )
# )
```

```python
# All-route weekly TS summary by DC_CD

ts_all_routes_df = (
    main_df.loc[
        main_df["DC_CD"].notna()
        & main_df["ts week"].notna()
        & main_df["CNSLDTN_ID"].notna()
        & main_df["ASN_TYPE_CD"].isin(["FM", "OV"])
        & pd.to_numeric(
            main_df["DELIVERY_SEQ"],
            errors="coerce"
        ).gt(0)
    ]
    .groupby(
        ["DC_CD", "ts week"],
        dropna=False
    )
    .agg(
        all_routes=(
            "ROUTE_ID",
            lambda x: sorted(
                x.dropna()
                .astype(str)
                .unique()
                .tolist()
            )
        ),
        route_count=(
            "ROUTE_ID",
            "nunique"
        ),
        total_distinct_stops=(
            "CNSLDTN_ID",
            "nunique"
        ),
        total_ts_charge=(
            "ts_rate_per_stop",
            lambda x: (
                x.dropna().iloc[0]
                if len(x.dropna()) > 0
                else 0.0
            )
        )
    )
    .reset_index()
)

# Recover total weekly TS charge:
# TS cost per stop x distinct CNSLDTN_ID count
ts_all_routes_df["total_ts_charge"] = (
    ts_all_routes_df["total_ts_charge"]
    * ts_all_routes_df["total_distinct_stops"]
)

# Weekly TS cost per distinct stop
ts_all_routes_df["ts_cost_per_stop"] = np.where(
    ts_all_routes_df["total_distinct_stops"].gt(0),
    (
        ts_all_routes_df["total_ts_charge"]
        / ts_all_routes_df["total_distinct_stops"]
    ),
    0.0
)

# Convert route list to a readable string
ts_all_routes_df["all_routes"] = (
    ts_all_routes_df["all_routes"]
    .apply(lambda routes: ", ".join(routes))
)

# Reorder and sort columns
ts_all_routes_df = (
    ts_all_routes_df[
        [
            "DC_CD",
            "ts week",
            # "all_routes",
            "route_count",
            "total_ts_charge",
            "total_distinct_stops",
            "ts_cost_per_stop"
        ]
    ]
    .sort_values(
        ["DC_CD", "ts week"]
    )
    .reset_index(drop=True)
)

display(ts_all_routes_df)
```

```python

# ============================================================
# DEBUG: RC + TS spreading for route 8162_G01 on May 11, 2026
# ============================================================

import pandas as pd
import numpy as np

TARGET_DC   = "XD_8162_OH_CLEVELAND.YZER"
TARGET_DATE = pd.Timestamp("2026-05-11").date()

# ---- Step 1: Build the FULL DC dataset for the whole week (needed for TS) ----
full_dc = lastmile_events.copy()
full_dc = full_dc[full_dc["ASN_TYPE_CD"].isin(['FM','RC','OV','TS'])]
full_dc = full_dc[full_dc["DELIVERY_TYPE"].isin(['REG'])]
full_dc = full_dc[full_dc["DC_CD"].str.upper().str.strip() == TARGET_DC.upper()]

full_dc["_date"] = pd.to_datetime(full_dc["DLVRY_ACTL_DATETIME"], errors="coerce").dt.date
full_dc["ts week"] = pd.to_datetime(full_dc["DLVRY_ACTL_DATETIME"], errors="coerce").dt.to_period("W-SUN").astype(str)
full_dc["ts key"] = full_dc["DC_CD"].astype(str) + "|" + full_dc["ts week"]

# TS total for the whole week at this DC
ts_total_dc = full_dc[full_dc["ASN_TYPE_CD"].eq("TS")].groupby("ts key")["TOPSIDE_COST"].sum()
ts_stop_cnt_dc = full_dc[full_dc["ASN_TYPE_CD"].isin(["FM","OV"])].groupby("ts key")["CNSLDTN_ID"].nunique()

print(f"TS rows at DC: {(full_dc['ASN_TYPE_CD']=='TS').sum()}")
print(f"TS total by week:\n{ts_total_dc}")
print(f"TS stop count by week:\n{ts_stop_cnt_dc}")
print()

# ---- Step 2: Build G01 + May 11 subset for RC spreading ----
dbg = full_dc[
    full_dc["ROUTE_ID"].str.contains("G01", na=False) &
    (full_dc["_date"] == TARGET_DATE)
].copy()

print(f"G01 rows on May 11: {len(dbg)}")
if len(dbg) == 0:
    print("No rows. Available dates for G01:")
    print(sorted(full_dc[full_dc["ROUTE_ID"].str.contains("G01", na=False)]["_date"].unique()))
else:
    dbg["helper column"] = (
        dbg["ROUTE_ID"].astype(str) + "|" +
        dbg["DC_CD"].astype(str) + "|" +
        pd.to_datetime(dbg["DLVRY_ACTL_DATETIME"]).dt.strftime("%Y%m%d")
    )

    stop_num2 = pd.to_numeric(dbg["DELIVERY_SEQ"], errors="coerce")
    dbg["stop helper"] = (
        dbg["helper column"] + "|" +
        np.where(stop_num2.fillna(0).eq(0), "RETURN", stop_num2.fillna(0).astype(int).astype(str))
    )
    dbg["dist stop flag"] = (~dbg["stop helper"].duplicated()).astype(int)

    rc_total_dbg = dbg[dbg["ASN_TYPE_CD"].eq("RC")].groupby("helper column")["TOTAL_RATE"].sum()
    fm_cnt_dbg   = dbg[(dbg["ASN_TYPE_CD"].eq("FM")) & (dbg["dist stop flag"].eq(1))].groupby("helper column").size()

    dbg["_rc_total"] = dbg["helper column"].map(rc_total_dbg).fillna(0)
    dbg["_fm_cnt"]   = dbg["helper column"].map(fm_cnt_dbg).fillna(0)
    dbg["_rc_share"] = np.where(dbg["_fm_cnt"] > 0, dbg["_rc_total"] / dbg["_fm_cnt"], 0)

    _has_rc = dbg["_rc_total"] > 0

    dbg["cost per stop"] = np.where(
        dbg["ASN_TYPE_CD"].isin(["RC", "TS"]), 0,
        np.where(
            ~_has_rc,
            np.where(dbg["ASN_TYPE_CD"].eq("OV"), dbg["LASTMILE_TOTAL_COST"], dbg["TOTAL_RATE"]),
            np.where(
                (dbg["ASN_TYPE_CD"].eq("FM")) & (dbg["dist stop flag"].eq(1)),
                dbg["TOTAL_RATE"] + dbg["_rc_share"],
                np.where(
                    (dbg["ASN_TYPE_CD"].eq("FM")) & (dbg["dist stop flag"].eq(0)),
                    0,
                    dbg["TOTAL_RATE"],
                ),
            ),
        ),
    )

    # ---- TS spreading using FULL DC week data (matching main cell logic) ----
    dbg["ts dist stop flag"] = (
        dbg["ASN_TYPE_CD"].isin(["FM","OV"])
    ) & (~dbg.duplicated(subset=["ts key","CNSLDTN_ID"]))
    dbg["ts dist stop flag"] = dbg["ts dist stop flag"].astype(int)

    dbg["_ts_total"] = dbg["ts key"].map(ts_total_dc).fillna(0)
    dbg["_ts_cnt"]   = dbg["ts key"].map(ts_stop_cnt_dc).fillna(0)
    dbg["_ts_share"] = np.where(dbg["_ts_cnt"] > 0, dbg["_ts_total"] / dbg["_ts_cnt"], 0)

    dbg["cost per stop"] = dbg["cost per stop"] + np.where(
        dbg["ts dist stop flag"].eq(1), dbg["_ts_share"], 0
    )

    dbg = dbg[dbg["ASN_TYPE_CD"].ne("TS")].copy()

    show_cols = ["ROUTE_ID", "CNSLDTN_ID", "DELIVERY_SEQ", "ASN_TYPE_CD",
                 "TOTAL_RATE", "_rc_share", "_ts_total", "_ts_cnt", "_ts_share", "cost per stop"]

    for route, grp in dbg.groupby("helper column"):
        rc = grp["_rc_total"].iloc[0]
        fm = int(grp["_fm_cnt"].iloc[0])
        paid = grp[grp["cost per stop"] > 0]
        print(f"\n{'='*60}")
        print(f"Route: {route}")
        print(f"  RC total: ${rc:.4f} | FM stops: {fm} | RC share: ${grp['_rc_share'].iloc[0]:.4f}")
        print(f"  TS total (DC week): ${grp['_ts_total'].iloc[0]:.4f} | TS stops: {grp['_ts_cnt'].iloc[0]} | TS share: ${grp['_ts_share'].iloc[0]:.4f}")
        print(f"  Mean CPS (paid): ${paid['cost per stop'].mean():.4f}")
        print()
        display(grp[show_cols].sort_values("DELIVERY_SEQ").reset_index(drop=True))
```

```python
# Extract day from Delivery ETA Start
# transvoyant_df['Delivery_Date'] = (
#     pd.to_datetime(
#         transvoyant_df['Delivery ETA Start']
#             .astype(str)
#             .str.replace(r'\\s+(MST|MDT|CST|CDT|PST|PDT)$', '', regex=True),
#         errors='coerce'
#     )
#     .dt.date
# )
# Extract day / year / month from Delivery ETA Start

_delivery_dt = pd.to_datetime(
    transvoyant_df['Delivery ETA Start']
        .astype(str)
        .str.replace(r'\\s+[A-Z]{2,4}$', '', regex=True),
    errors='coerce'
)

transvoyant_df['Delivery_Date'] = _delivery_dt.dt.date
transvoyant_df['year'] = _delivery_dt.dt.year
transvoyant_df['month'] = _delivery_dt.dt.month

# Max distance and stop count per Route / Day

transvoyant_max_dist = (
    transvoyant_df.groupby(
        [
            'Fill DC',
            'XDOCK',
            'Delivery Route ID',
            'Delivery_Date',
            'year',
            'month'
        ],
        as_index=False
    )
    .agg(
        total_miles=('Total distance traveled, mi', 'max'),
        stop_count=('Delivery Stop ID', 'nunique')
    )
)

# Roll up to XDock / Day

total_xd_daily_miles = (
    transvoyant_max_dist
    .groupby(
        [
            'Fill DC',
            'XDOCK',
            'Delivery_Date',
            'year',
            'month'
        ],
        as_index=False
    )
    .agg(
        total_miles=('total_miles', 'sum'),
        stop_count=('stop_count', 'sum')
    )
    .assign(
        miles_per_stop=lambda df: df['total_miles'] / df['stop_count']
    )
)

# Average miles per stop by XDock / Year / Month

xd_avg_daily_miles_per_stop = (
    total_xd_daily_miles
    .groupby(
        ['XDOCK', 'year', 'month'],
        as_index=False
    )
    .agg(
        avg_miles_per_stop=('miles_per_stop', 'mean')
    )
)

# Monthly baseline (median across XDocks)

xd_avg_daily_miles_per_stop['baseline_miles_per_stop'] = (
    xd_avg_daily_miles_per_stop
    .groupby(['year', 'month'])['avg_miles_per_stop']
    .transform('median')
)

print(
    xd_avg_daily_miles_per_stop[
        ['year', 'month', 'baseline_miles_per_stop']
    ].drop_duplicates()
)

# Multiplier

xd_avg_daily_miles_per_stop['miles_per_stop_norm_multiplier'] = (
    xd_avg_daily_miles_per_stop['avg_miles_per_stop']
    / xd_avg_daily_miles_per_stop['baseline_miles_per_stop']
)

# Quantile clipping within month

lower = (
    xd_avg_daily_miles_per_stop
    .groupby(['year', 'month'])['miles_per_stop_norm_multiplier']
    .transform(lambda x: x.quantile(0.10))
)

upper = (
    xd_avg_daily_miles_per_stop
    .groupby(['year', 'month'])['miles_per_stop_norm_multiplier']
    .transform(lambda x: x.quantile(0.90))
)

xd_avg_daily_miles_per_stop['miles_per_stop_norm_multiplier_qt'] = (
    xd_avg_daily_miles_per_stop['miles_per_stop_norm_multiplier']
    .clip(lower=lower, upper=upper)
)

display(xd_avg_daily_miles_per_stop)


# display(xd_avg_daily_miles_per_stop)
# total_xd_daily_miles['Fill DC']=total_xd_daily_miles['Fill DC'].astype("str")
```

```python
# ============================================================
# CELL 8: FIT EXPECTED COST MODELS
# ============================================================

def make_preprocessor(numeric_features, categorical_features):
    numeric_pipe = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="median")),
        ]
    )

    categorical_pipe = Pipeline(
        steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            (
                "encoder",
                OrdinalEncoder(
                    handle_unknown="use_encoded_value",
                    unknown_value=-1,
                ),
            ),
        ]
    )

    return ColumnTransformer(
        transformers=[
            ("num", numeric_pipe, numeric_features),
            ("cat", categorical_pipe, categorical_features),
        ],
        remainder="drop",
    )


def make_quantile_model(q):
    return Pipeline(
        steps=[
            (
                "prep",
                make_preprocessor(
                    numeric_features,
                    categorical_features,
                ),
            ),
            (
                "model",
                HistGradientBoostingRegressor(
                    loss="quantile",
                    quantile=q,
                    max_iter=180,
                    learning_rate=0.06,
                    max_leaf_nodes=31,
                    l2_regularization=0.05,
                    random_state=RANDOM_STATE,
                ),
            ),
        ]
    )


if FIT_FULL_QUANTILES:
    quantiles = [0.10, 0.25, 0.50, 0.75, 0.90]
else:
    quantiles = [0.10, 0.50, 0.90]

X_train, X_test, y_train, y_test, w_train, w_test = train_test_split(
    X,
    y,
    weights,
    test_size=0.20,
    random_state=RANDOM_STATE,

models = {}

for qv in quantiles:
    print("Fitting quantile:", qv)

    model = make_quantile_model(qv)

    model.fit(
        X_train,
        y_train,
        model__sample_weight=w_train,
    )

    models[qv] = model

    gc.collect()

pred_p50_log = models[0.50].predict(X_test)
pred_p50_cost = np.expm1(pred_p50_log).clip(min=0)
y_test_cost = np.expm1(y_test)

model_metrics = {
    "model_groups": int(len(model_df)),
    "feature_count": int(len(feature_cols)),
    "quantiles_fit": str(quantiles),
    "mae_cost": float(mean_absolute_error(y_test_cost, pred_p50_cost)),
    "median_abs_error_cost": float(median_absolute_error(y_test_cost, pred_p50_cost)),
    "r2_log_target": float(r2_score(y_test, pred_p50_log)),
}

display(pd.DataFrame([model_metrics]))

del X_train, X_test, y_train, y_test, w_train, w_test
gc.collect()
```

We train a quantile gradient boosting model on aggregated market-level operational characteristics to estimate the expected cost-per-stop range (P10, P50, P90). Markets are then evaluated based on how far their actual median cost deviates from the expected range, rather than relying solely on normalized cost multipliers. This provides a statistically grounded assessment of potential overpay and underpay conditions.

```python
# ============================================================
# CELL 9: SCORE ALL MARKET GROUPS
# ============================================================

scorecard = market_df.copy()

score_eligible = (
    scorecard["actual_median_cps"].notna()
    & (scorecard["actual_median_cps"] > 0)
)

for qv, model in models.items():
    col = f"pred_p{int(qv * 100):02d}"

    scorecard[col] = np.nan

    scorecard.loc[score_eligible, col] = np.expm1(
        model.predict(scorecard.loc[score_eligible, feature_cols])
    ).clip(min=0)

# Add missing quantile columns if simplified mode
if "pred_p25" not in scorecard.columns:
    scorecard["pred_p25"] = np.nan

if "pred_p75" not in scorecard.columns:
    scorecard["pred_p75"] = np.nan

# Enforce monotonic quantiles
if FIT_FULL_QUANTILES:
    qcols = ["pred_p10", "pred_p25", "pred_p50", "pred_p75", "pred_p90"]
else:
    qcols = ["pred_p10", "pred_p50", "pred_p90"]

arr = scorecard[qcols].to_numpy()
arr = np.sort(arr, axis=1)
scorecard[qcols] = arr

scorecard["model_residual"] = (
    scorecard["actual_median_cps"]
    - scorecard["pred_p50"]
)

scorecard["model_residual_pct"] = safe_div(
    scorecard["actual_median_cps"],
    scorecard["pred_p50"],
) - 1

if "median_normalized_cost" in scorecard.columns:
    scorecard["normalization_sensitivity_pct"] = safe_div(
        scorecard["median_normalized_cost"],
        scorecard["actual_median_cps"],
    ) - 1
else:
    scorecard["normalization_sensitivity_pct"] = np.nan

scorecard["norm_sensitivity_abs_pct"] = np.abs(
    scorecard["normalization_sensitivity_pct"]
)

if "median_cleansheet_cost_per_stop_conservative" in scorecard.columns:
    scorecard["cleansheet_cons_gap_pct"] = safe_div(
        scorecard["actual_median_cps"],
        scorecard["median_cleansheet_cost_per_stop_conservative"],
    ) - 1
else:
    scorecard["cleansheet_cons_gap_pct"] = np.nan

if "median_cleansheet_cost_per_stop_aggressive" in scorecard.columns:
    scorecard["cleansheet_aggr_gap_pct"] = safe_div(
        scorecard["actual_median_cps"],
        scorecard["median_cleansheet_cost_per_stop_aggressive"],
    ) - 1
else:
    scorecard["cleansheet_aggr_gap_pct"] = np.nan

display(scorecard.head())
```

```python
# ============================================================
# CELL 10: CLASSIFICATION LOGIC
# ============================================================

scorecard["confidence_score"] = 0

scorecard.loc[
    scorecard["paid_records"] >= MIN_GROUP_N,
    "confidence_score",
] += 1

scorecard.loc[
    scorecard["paid_records"] >= MIN_GROUP_N * 3,
    "confidence_score",
] += 1

scorecard.loc[
    "confidence_score",
] += 1

scorecard.loc[
    scorecard["norm_sensitivity_abs_pct"].fillna(0) <= 0.40,
    "confidence_score",
] += 1

scorecard["confidence"] = np.select(
    [
        scorecard["confidence_score"] >= 3,
        scorecard["confidence_score"] == 1,
        scorecard["confidence_score"] == 1,
    ],
    [
        "High",
        "Medium",
        "Low",
    ],
    default="Very Low",
)

scorecard["classification"] = "Normal / inside expected band"

strong_conf = scorecard["confidence"].isin(["High", "Medium"])

if FIT_FULL_QUANTILES:
    overpay_possible_condition = (
        (scorecard["model_residual_pct"] >= OVERPAY_POSSIBLE)
        & (scorecard["actual_median_cps"] > scorecard["pred_p75"])
    )

    underpay_possible_condition = (
        (scorecard["model_residual_pct"] <= UNDERPAY_POSSIBLE)
        & (scorecard["actual_median_cps"] < scorecard["pred_p25"])
    )
else:
    overpay_possible_condition = (
        scorecard["model_residual_pct"] >= OVERPAY_POSSIBLE
    )

    underpay_possible_condition = (
        scorecard["model_residual_pct"] <= UNDERPAY_POSSIBLE
    )

scorecard.loc[
    (scorecard["model_residual_pct"] >= OVERPAY_STRONG)
    & (scorecard["actual_median_cps"] > scorecard["pred_p90"])
    & strong_conf,
    "classification",
] = "Strong overpay candidate"

scorecard.loc[
    scorecard["classification"].eq("Normal / inside expected band")
    & overpay_possible_condition,
    "classification",
] = "Possible overpay"

scorecard.loc[
    (scorecard["model_residual_pct"] <= UNDERPAY_STRONG)
    & (scorecard["actual_median_cps"] < scorecard["pred_p10"])
    & strong_conf,
    "classification",
] = "Strong underpay candidate"

scorecard.loc[
    scorecard["classification"].eq("Normal / inside expected band")
    & underpay_possible_condition,
    "classification",
] = "Possible underpay"

scorecard.loc[
    (scorecard["paid_records"] < MIN_GROUP_N)
    | scorecard["confidence"].eq("Very Low")
    | scorecard["actual_median_cps"].isna()
    | scorecard["pred_p50"].isna(),
    "classification",
] = "Not enough evidence"

scorecard["cleansheet_overpay_support"] = (
    scorecard["cleansheet_aggr_gap_pct"].notna()
    & (scorecard["cleansheet_aggr_gap_pct"] > 0)
)

scorecard["cleansheet_underpay_support"] = (
    scorecard["cleansheet_cons_gap_pct"].notna()
    & (scorecard["cleansheet_cons_gap_pct"] < 0)
)

scorecard["business_note"] = np.select(
    [
        scorecard["classification"].str.contains("overpay", case=False, na=False)
        & scorecard["cleansheet_overpay_support"],

        scorecard["classification"].str.contains("underpay", case=False, na=False)
        & scorecard["cleansheet_underpay_support"],

        scorecard["classification"].eq("Not enough evidence"),

        scorecard["norm_sensitivity_abs_pct"].fillna(0) > 0.40,
    ],
    [
        "Model and cleansheet both point high",
        "Model and cleansheet both point low",
        "Low volume or weak basis",
        "Sensitive to multiplier assumptions",
    ],
    default="Model-based residual signal",
)

scorecard = scorecard.sort_values(
    [
        "classification",
        "model_residual_pct",
        "paid_records",
    ],
    ascending=[
        True,
        False,
        False,
    ],
)

display(
    scorecard["classification"]
    .value_counts()
    .rename_axis("classification")
    .reset_index(name="count")
)

display(scorecard.head(25))
```
